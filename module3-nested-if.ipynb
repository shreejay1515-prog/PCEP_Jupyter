```python
# Nested if

answer_a = input('Do you like travelling y/n: ')
if answer_a == 'y':
    answer_b = input('Do you like Asia y/n:')
    if answer_b == 'y':
        print('Excellent! You win trip to Thailand')
    else:
        print('Sorry to hear that')
else:
    print('Sorry to hear that')
```

```python
# Coding Exercise 5 - Refund Policy Helper - my code 

days_purchased = int(input('How many days ago have you purchased the item? '))
item_used = input('Have you used the item at all [y/n]? ')
item_faulty = input('Has the item broken down on its own [y/n]? ')

if (days_purchased <= 10 and item_used == 'n') or item_faulty == 'y':
    print('You can get a refund.')
else:
    print('You cannot get a refund.')
```
