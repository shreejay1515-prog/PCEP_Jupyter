```python
# Each element has an index 
# List indices start from 0 not from 1.

top_cities = ['New York','Los Angeles','Chicago','Houston','Phoenix']
#indices -> [0 , 1, 2 ,3 , 4]
print(top_cities)
```

```python
# -ve indices to retrieve value is also possible 
# -1 - will give you last element of list
# Mark how it returns apostrophe as well.

top_cities[-1]
```

```python
top_cities[-5]
```

```python
top_cities[-6]
```

```python
# we can also take multiple elements together - with help of slicing [a:b] 
# we do not take last element only 0 & 1. Element at index 2 is exclusive.

top_cities[0:2]
```

```python
# take elements from specific position to end

top_cities = ['New York','Los Angeles','Chicago','Houston','Phoenix']
top_cities[2:]
```

```python
# take elements from start to specific position - first 4 elements

top_cities[:4]
```

```python
top_cities [:]

#same as top_cities (refer below cell)
```
