```python
# Section 4: loops exercise

release_year = int(input('When was Python 1.0 released? '))
while release_year != 1994:
    if release_year > 1994:
        print('It was earlier than that!')
        release_year = int(input('When was Python 1.0 released? '))
    if release_year < 1994:
        print('It was later than that!')
        release_year = int(input('When was Python 1.0 released? '))
print('Correct! ')
```
