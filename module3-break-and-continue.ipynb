```python
# break immediately exits loop - used inside if loop
# infinite loop where statement is always true

while True: 
    name = input('Enter your name or Exit to quit the program')
    if name == 'Exit':
        break
    print('Hello',name)
```

```python
# continue - exits current iteration and enters next one
# used with if staement when we want to break certain iteration

# skip the iteration for numbers divided by 5

for i in range (1,21):
    if i % 5 == 0:
        continue
    print(i)
print('Finished')
```
