```python
# available logical operators
#
# < less than
# > greater than
# <= less than or equal to
# >= greater than or equal to 
# == equals
# != not equals
```

```python
password = input('Do you know the password?')
if password != '--secret':
    print ('not correct!')
else:
    print ('correct password!')
```

```python
# When python sees a logical operator it always checks for a boolean value
True   - T should be capital
False  - F should be capital
```

```python
if True:
    print('condition met')
```

```python
if False:
    print('condition met')
```

```python
# checking for int and float equal to

if 1 == 1:
    print('True')
```

```python
if 2 == 1:
    print('True')
```

```python
#since 2 != 1 we still get no output

if 2 == 1:
    print('False')
```

```python
# string & float - python automatically verifies across data types

if 2.0 == 2:
    print('True')
```

```python
if 3 == 3.0:
    print('True')
```
