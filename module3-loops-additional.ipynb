```python
# Nested Loops

for a in range (1,6):
    for b in range (1,6):
        print(a ,'x', b, '=', a * b)
```

```python
# Remember - Else branch is always executed atleast once (exception: break statement)

# Loops with Else statement - 1

i = 5
while i < 5:
    print('i')
    i += 1
else:
    print ('else:', i)
```

```python
# Loops with Else statement - 2

i = 2
while i < 5:
    print(i)
    i += 1
else:
    print ('else:', i)
```
