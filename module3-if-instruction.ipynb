```python
# conditional statement - python

user_age = int(input('What is your age?'))

# if block
if user_age > 30:
    # conditions mentioned in if statement must be indented - exactly 4 space
    print('You are over 30 years of age')
    print('Great,you qualify!')
    
#else-if block - multiple elif can be used in python
elif user_age == 30: # == is usedd
    print('You are over 30 years of age')
    print('Great,you qualify on brim!')
    
# else block
else:
    print('You are 30 years or younger', end='-')
    print('You do not qualify') 
```

```python
Use of = & == in python:

Single =  : assigning value to variables
Double == : checking equality
```
