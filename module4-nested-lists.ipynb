```python
cells = [['A1','A2','A3'],['B1','B2','B3']]
for x in cells:
    print('Element:', x)
```

```python
cells = [['A1','A2','A3'],['B1','B2','B3']]
for x in cells:
    for y in x:
        print('Element:', y)
```

```python
table = [['A1','A2','A3'],['B1','B2','B3']]
for row in table:
    for cell in row:
        print('Element:', cell)
```

```python
# Paste elements in such a way that it resembles the table

table = [['A1','A2','A3'],['B1','B2','B3']]
for row in table:
    for cell in row:
        print(cell,'', end='')
    print()
```

```python
# nested list comprehension - creating a table with identical loops
# no limit to dimensions - lists can be 3, 4  or more dimensional

table = [[i for i in range (1,6)] for j in range (4)]
print(table)
```
