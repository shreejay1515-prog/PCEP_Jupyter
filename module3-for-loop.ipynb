```python
# FOR LOOP- in python is used for going through all the elements of specific sequence 
# sequences - are special data structure which store more than one value and browse sequentially - element by element
# example of sequence : string 'hello' - string contains multiple characters

# no need to check condition like while loop & increment counter - python self increments
```

```python
# for loop syntax : 
# for variable in 'string':
#    action ex: print

for letter in 'hello!':
    print('current letter:', letter)
```

```python
# scanning or browsing sequence is called iterating a sequence
# running body/loop once: iteration 
```

```python
# while loop counter

counter = 1
while counter < 11:
    print(counter)
    counter += 1
print('Finished!')
```

```python
# for loop - counter in range (1,11) - does not include 11
# syntax: for variable in range (1 - start value ,x - stop value):
# range function is special function which generates all desired integer value for desired variable.

for counter in range(1,11):
    print (counter)
print ('Finished')
```

```python
# Alternate syntax 

for counter in range(11):
    print (counter)
print ('Finished')
```

```python
# for iterating with gap/jump of 2 

for counter in range(1,11,2):
    print (counter)
print ('Finished')
```

```python
# for iterating with gap/jump of 4 

for counter in range(1,11,4):
    print (counter)
print ('Finished')
```
