```python
#learnings / objectives :

swapping values
method: sort(), sort(reverse=True) #method because it changes data
list_name.sort() : sorts original data list - data gets changed

List function: sorted()
soted(list_name) :gives back a new, sorted list, keeps the original unchecked
```

```python
first = input('Enter first number: ')
second = input('Enter second number: ')
print('before swapping:', first, second)
first = second # first value is lost
second = first
print('after swapping:',first,second)
```

```python
# saving first value to temporary second variable so it does not gets lost

first = input('Enter first number: ')
second = input('Enter second number: ')
print('before swapping:', first, second)
temporary = first
first = second
second = temporary
print('after swapping:',first,second)
```

```python
#swapping logic for variables var1, var2 = var2, var1 to exchange values 

first = input('Enter first number: ')
second = input('Enter second number: ')
print('before swapping:', first, second)
first, second = second, first 
print('after swapping:',first,second)
```

```python
# switch values of first & last element in list element[0],element[4] = element[4],element[0]

top_cities = ['New York City','Los Angeles','Chicago','Houston','Phoenix']
top_cities[0], top_cities[4] = top_cities[4], top_cities[0]
print(top_cities)
```

```python
# new method sort()
# sort can be used to arrange int, str & float data types

top_cities = ['New York City','Los Angeles','Chicago','Houston','Phoenix']
top_cities.sort()
print(top_cities)
```

```python
random_numbers = [2, 3, 5, -4, 0, 1]
random_numbers.sort()
print(random_numbers)
```

```python
random_numbers = [2, 3, 5, -4, 0, 1]
random_numbers.sort(reverse = True)
print(random_numbers)
```
