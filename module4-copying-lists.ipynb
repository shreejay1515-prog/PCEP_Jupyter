```python
learning/objective:

1. How to copy a list in new variable
2. Name given to the list is not name of the list, however its the location where list is stored known as referencing/(references)
3. How slicer can be used to copy old list to new list and having 2 different lists [:]
```

```python
# notice how we updated original, but this did not change name_new - same rule is applied to int, float and boolean
# however, not true for data collections like lists

name_original = 'John Snow'
name_new = name_original
name_original = 'Denarys Targyeren'

print(name_original, name_new)
```

```python
# we are trying to change list completely but we are just referencing - point at memory location where list is stored
# IMPORTANT: Note how changing the first element also results in changing of list_original

list_original = [1, 2, 3]
list_new = list_original
list_original[0] = -5     # changing element 1
print('original:', list_original, '\nnew:',list_new)
```

```python
# use slicing "[:]" to retain original list
# In output new list is retained - just original name in 'Jon Snow' code
# modifying one of them also modifies other

list_original = [1, 2, 3]
list_new = list_original [:]
list_original[0] = -5     # changing element 1
print('original:', list_original, '\nnew:',list_new)
```

```python
list_original = [1, 2, 3]
list_new = list_original [:2]
list_original[0] = -5     
print('original:', list_original, '\nnew:',list_new)
```

```python
# list_original and list_new point at same list - if one is modified other is also modified.

list_original = [1, 2, 3]
list_new = list_original
```

```python
# copy a list in another variable and have 2 different lists in 2 different variables
# use slicers- [:]

list_original = [1, 2, 3]
list_new = list_original[:]
```
