```python
 3 things we have covered so far 

 1. function - syntax -> function name () - ex: print()
 2. instruction - syntax -> instruction list_name[1] / [1:2] etc - ex: del top_cities[2]
 3. method - syntax -> list_name.method('positional reference') - ex: top_cities.append(3),top_cities.insert(1,10)
```

```python
Methods: 

Methods are Function that 'belong' to specific data.
```

```python
# Append
# syntax - list_name.append (value) - new value added at the end

book_ratings = [7 ,9 ,5 ,6 ,8]
book_ratings.append (4)
print (book_ratings)
```

```python
# only using append (method) without list name gives error
append (4)
```

```python
# insert
# syntax - list_name.insert(position, value) - new value can be added at any position

book_ratings = [7 ,9 ,5 ,6 ,8]
book_ratings.insert (1,10)
print(book_ratings)
```
