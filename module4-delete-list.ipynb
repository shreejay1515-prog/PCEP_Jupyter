```python
#'del' -  2 actions: 1.del the cities and 2.shift the other elements to left 

top_cities = ['New York','Los Angeles','Singapore','Chicago','Houston','Phoenix']
#indices -> [0 , 1, 2 ,3 , 4]
del top_cities[2]
print (top_cities)
```

```python
# 3.new element 2
top_cities[2]
```

```python
#3 del - is not funtion call, it's instruction - because function requires name (followed by brackets) - print('content')

top_cities = ['New York','Los Angeles','Singapore','Chicago','Houston','Phoenix']
del top_cities[3:]
print (top_cities)
```

```python
# This prompt with [:] will give an empty list

top_cities = ['New York','Los Angeles','Singapore','Chicago','Houston','Phoenix']
del top_cities[:]
print (top_cities)
```
