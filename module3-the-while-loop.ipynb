```python
# while condition - unless condition is met python/program will not leave the loop.

while condition:
    do_something()
    do_something2()
```

```python
# print 1 to 10 - while condition is satisfied

counter = 1
while counter < 11:
    print(counter)
    counter += 1
print ("Finished!")
```

```python
# print 1 to 10 - while condition is NOT satisfied at the beginning - then python will not even run body once

counter = 1
while counter > 11:
    print(counter)
    counter += 1
print ("Finished!")
```

```python
# print ''' ... ''' for multi lines text print

secret_number = 14
print ('''
==========================
=== SECRET NUMBER GAME ===
==========================
''')

guess_number = int(input('Try to guess number between 1 to 20: '))
while guess_number != secret_number:
    print('wrong')
    guess_number = int(input('Try to guess number between 1 to 20: '))
print('Perfect! You have guessed right number')
    
```
