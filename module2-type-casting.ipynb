```python
#Type Casting: Changing data type from one to another is called type casting.
#ex: float to str, int to str etc 

height_cm = input('Height in cm:')
float_height = float(height_cm)
print('Your height in feet is:',float_height/30.48)
```

```python
# convert height to cm

height_cm = float(input('Height converter: Enter your height in cm:'))
print('Your height in feet is:',height_cm/30.48)
```

```python
temp_c = input('Enter the temperature today in celcius degree:')
temp_f = float(temp_c) * 1.8 + 32
temp_statement = str (temp_c) + ' degree celc is ' + str(temp_f) + ' degree far'
print(temp_statement)
```

```python
# type casting test

work_hours_month = float(input('How many hours did you work last month? '))
hourly_rate = float(input('What is your hourly rate? '))
salary = work_hours_month * hourly_rate
print('Last month, you earned', salary,'dollars')
```
