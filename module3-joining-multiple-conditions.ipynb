```python
# Run code when at least one or multiple conditions are met

# and

user_age = int(input('What is your age? '))
user_country = input('What is your country? ')

if user_age < 25 and user_country == 'Germany':
    print('You can apply for German exchange program')
else:
    print('You do not qualify.')
```

```python
# or

user_country = input("What is your country? ")

if user_country == 'Norway' or user_country == 'Sweden' or user_country == 'Denmark':
    print('You can apply for Scandinavian visa')
else:
    print('You cannot apply')
```

```python
# not
# not negates boolean value, if condition is True it will turn it False 

user_country = input("What is your country? ")

if not user_country == 'Norway':
    print('You are not from Norway')
else:
    print('You are from Norway')
```

```python
# merging all the operators 
# use '\' in middle of line for line break and continue on next line
# it flows from left to right - checks for 'not with and' later checks for 'or'

user_age = int(input('What is your age ?'))
user_country = input('What is your country?')

if not user_country == 'Germany' and user_age < 23 or \
               user_country == 'Germany' and user_age < 25:
    print('You qualify!')
else:
    print ('You do not qualify!')
```

```python
# above example same result - with brackets

user_age = int(input('What is your age ?'))
user_country = input('What is your country?')

if ((not user_country == 'Germany') and user_age < 23) or \
               user_country == 'Germany' and user_age < 25:
    print('You qualify!')
else:
    print ('You do not qualify!')
```
