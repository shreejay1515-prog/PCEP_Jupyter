```python
learning/objective:

1. in operator - can be used for checking letters in words or elements as whole in lists
2. in orperator can be used with *if to directly check condition
3. not in - reverse of in
```

```python
# in operator for checking characters

for char in 'happy marriage':
    print(char)
```

```python
# in operator for checking elements in the list 
# in operator for directly checking condition

invited_guests = ['Kate', 'Adam', 'Kerry', 'Joe', 'Ann', 'Mary']
name = input('name of the guest:') 
if name in invited_guests:
    print ('You are welcome!')
else:
    print ('You are not invited')
```

```python
# 'not in' operator for directly checking condition

invited_guests = ['Kate', 'Adam', 'Kerry', 'Joe', 'Ann', 'Mary']
name = input('name of the guest:') 
if name not in invited_guests:
    print ('You are Welcome!')
else:
    print ('You are not invited')
```
